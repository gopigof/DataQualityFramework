```python
import pandas as pd
import numpy as np
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler
from sklearn.impute import SimpleImputer

# Load the dataset
df = pd.read_excel('/Users/shruti/Downloads/Assignment 1_7374/Retail_Data1.xlsx')

# Handling outliers with new column for 'Cost'
Q1 = df['Cost'].quantile(0.25)
Q3 = df['Cost'].quantile(0.75)
IQR = Q3 - Q1
df['Cost_Capped'] = np.where(df['Cost'] > (Q3 + 1.5 * IQR), Q3 + 1.5 * IQR, df['Cost'])

# Normalize 'Cost' in a new column (assuming 'Cost' is the intended column)
scaler = MinMaxScaler()
df['Cost Normalized'] = scaler.fit_transform(df[['Cost']])

# Standardize 'Profit Margin' in a new column
standard_scaler = StandardScaler()
df['Profit Margin Standardized'] = standard_scaler.fit_transform(df[['Profit_Margin']])

# Encoding categorical variables with new columns for each category
#'Customer Region' is categorical
if 'Customer Region' in df.columns:
    df['Customer Region Encoded'] = pd.Categorical(df['Customer Region']).codes

# Define age bins and labels
bins = [0, 18, 25, 35, 45, 55, 65, 100]
labels = ['0-18', '18-24', '25-34', '35-44', '45-54', '55-64', '65-100']

# Convert 'Customer_Age_Group' to categorical age groups and save in a new column
df['Customer_Age_Group_Grouped'] = pd.cut(df['Customer_Age_Group'], bins=bins, labels=labels, right=False)

# Log transformation for 'Cost' to handle skewness in a new column (assuming 'Cost' is the intended column)
# TODO: Remove
df['Log Cost'] = np.log1p(df['Cost'])

# Handling missing values in numeric columns using median imputation
imputer = SimpleImputer(strategy='median')
numeric_columns = df.select_dtypes(include=[np.number]).columns
df[numeric_columns] = imputer.fit_transform(df[numeric_columns])

df['Year'] = df['Year'].fillna(0).astype(int)
df['Month'] = df['Month'].fillna(0).astype(int)
df['Day'] = df['Day'].fillna(0).astype(int)
df['Hour'] = df['Hour'].fillna(0).astype(int)

# Correct formatting of the datetime string to match the format specification
df['Datetime'] = pd.to_datetime(df['Year'].astype(str) + '-' +
                                df['Month'].astype(str).str.zfill(2) + '-' +
                                df['Day'].astype(str).str.zfill(2) + ' ' +
                                df['Hour'].astype(str).str.zfill(2),
                                format='%Y-%m-%d %H')

# Sum sales data across product categories and create a new column
df['Total_Electronics_Sales'] = df[['Electronics_Laptops', 'Electronics_Smartphones', 'Electronics_Headphones']].sum(axis=1)

# Encode 'Customer_Region' and save in a new column
df['Customer_Region_Encoded'] = pd.Categorical(df['Customer_Region']).codes

# Remove conditionally the Marketing and Profit
# TODO: Calculate ROI from marketing spend and save in a new column
df['ROI_From_Marketing_Spend'] = df['Profit_Margin'] / df['Marketing_Spend']

# Normalize 'Profit Margin' and save in a new column
scaler = MinMaxScaler()
df['Normalized_Profit_Margin'] = scaler.fit_transform(df[['Profit_Margin']])

# Standardize 'Marketing Spend' and save in a new column
standard_scaler = StandardScaler()
df['Standardized_Marketing_Spend'] = standard_scaler.fit_transform(df[['Marketing_Spend']])


# Save the cleaned and transformed dataset
df.to_excel('/Users/shruti/Downloads/cleaned_Retail_Data1.xlsx', index=False)
```
